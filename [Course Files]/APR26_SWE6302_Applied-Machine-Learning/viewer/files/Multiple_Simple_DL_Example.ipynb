# **sparse_categorical_crossentropy**

This code will train a simple neural network to classify MNIST handwritten digits. The network has two hidden layers, each with 128 neurons and 10 neurons respectively. The output layer has 10 neurons, one for each digit (class).

Note: Use Google Colab Notebook to run the code if possible

````python
#Import Tensorflow Library and Matplotlib Library
import tensorflow as tf
````

````python
# Load the MNIST dataset.
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()
````

````python
# Convert the data to float32 format.
x_train = tf.cast(x_train, tf.float32) / 255.0
x_test = tf.cast(x_test, tf.float32) / 255.0
````

Updated DL model's diagramatic representation as follows:

```
Input (28x28)
|
Flatten
|
Dropout (20%)
|
Dense (128, relu)
|
Dropout (20%)
|
Dense (64, relu)
|
Dense (10, softmax)
|
Output
```



````python
# Create a sequential model. This is a slightly updated DL model with dropout layers
model = tf.keras.Sequential([
  tf.keras.layers.Flatten(input_shape=(28, 28)),

  # Add a dropout layer with a dropout rate of 20%.
  tf.keras.layers.Dropout(0.2),

  # Add a hidden layer with 128 neurons.
  tf.keras.layers.Dense(128, activation='relu'),

  # Add another dropout layer with a dropout rate of 20%.
  tf.keras.layers.Dropout(0.2),

  # Add a hidden layer with 64 neurons.
  tf.keras.layers.Dense(64, activation='relu'),

  # Add a final hidden layer with 10 neurons and a softmax activation for classification.
  tf.keras.layers.Dense(10, activation='softmax') #You can also use a different activation here such as activation='sigmoid'
])
````

````python
# Compile the model.
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])
'''
You can also try out different loss functions as follows to see how the loss changes such as:
loss='binary_crossentropy' OR
loss='categorical_crossentropy' OR
loss=focal_loss(gamma=2) #Focal loss
'''

'''
#By default optimizer used in Keras here, which is the Adam optimizer, has a default learning rate of 0.001. But the learning rate can be specified as follows.
model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.005),
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])
'''
````

Note: ^^^^This loss function (sparse_categorical_crossentropy) is designed to be used with sparse labels, which means that it does not expect the labels to be one-hot encoded. In one-hot encoded label, each label is represented by a vector of 10 elements, with the ith element being 1 and the rest of the elements being 0.

````python
# Train the model.
model.fit(x_train, y_train, epochs=10)
````

````python
# Evaluate the model on the test set.
test_loss, test_acc = model.evaluate(x_test, y_test)
print('Test accuracy:', test_acc)
````

# **categorical_crossentropy**

````python
#Import Tensorflow Library and Matplotlib Library
import tensorflow as tf
from tensorflow.keras.utils import to_categorical
````

````python
# Convert the labels to one-hot encoded format.
y_train = to_categorical(y_train)
y_test = to_categorical(y_test)
````

````python
# Create a sequential model. This is a slightly updated DL model with dropout layers
model = tf.keras.Sequential([
  tf.keras.layers.Flatten(input_shape=(28, 28)),

  # Add a dropout layer with a dropout rate of 20%.
  tf.keras.layers.Dropout(0.2),

  # Add a hidden layer with 128 neurons.
  tf.keras.layers.Dense(128, activation='relu'),

  # Add another dropout layer with a dropout rate of 20%.
  tf.keras.layers.Dropout(0.2),

  # Add a hidden layer with 64 neurons.
  tf.keras.layers.Dense(64, activation='relu'),

  # Add a final hidden layer with 10 neurons and a softmax activation for classification.
  tf.keras.layers.Dense(10, activation='softmax') #You can also use a different activation here such as activation='sigmoid'
])
````

````python
# Compile the model.
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])
'''
You can also try out different loss functions as follows to see how the loss changes such as:
loss='binary_crossentropy' OR
loss='sparse_categorical_crossentropy' OR
loss=focal_loss(gamma=2) #Focal loss
'''

'''
#By default optimizer used in Keras here, which is the Adam optimizer, has a default learning rate of 0.001. But the learning rate can be specified as follows.
model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.005),
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])
'''
````

Note: ^^^^ categorical_crossentropy needs one-hot encoded labels.
Example:

```
Category | One-hot encoded vector
------- | --------
0      | [1, 0, 0]
1      | [0, 1, 0]
2     | [0, 0, 1]
```



````python
# Train the model.
model.fit(x_train, y_train, epochs=10)
````

````python
# Evaluate the model on the test set.
test_loss, test_acc = model.evaluate(x_test, y_test)
print('Test accuracy:', test_acc)
````

````python
# Load the MNIST dataset.
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()
````

````python
# Create a sequential model. This is a slightly updated DL model with dropout layers
model = tf.keras.Sequential([
  tf.keras.layers.Flatten(input_shape=(28, 28)),

  # Add a dropout layer with a dropout rate of 20%.
  tf.keras.layers.Dropout(0.2),

  # Add a hidden layer with 128 neurons.
  tf.keras.layers.Dense(128, activation='relu'),

  # Add another dropout layer with a dropout rate of 20%.
  tf.keras.layers.Dropout(0.2),

  # Add a hidden layer with 64 neurons.
  tf.keras.layers.Dense(64, activation='relu'),

  # Add a final hidden layer with 10 neurons and a softmax activation for classification.
  tf.keras.layers.Dense(10, activation='softmax') #You can also use a different activation here such as activation='sigmoid'
])
````

Note: ^^^^ Focal loss is a loss function used for classification tasks. It is designed to address the problem of class imbalance, which occurs when there are a lot more examples of one class than another.

Focal loss works by down-weighting the loss for easy examples and focusing on the loss for hard examples. This is done by applying a modulating factor to the cross-entropy loss function. The modulating factor is calculated as follows:


```
modulating_factor = (1 - p_t)^γ
```
where:

p_t is the predicted probability of the correct class
γ is a hyperparameter that controls the focus on hard examples
The value of γ determines how much focal loss down-weights easy examples. A higher value of γ will down-weight easy examples more heavily and focus on hard examples more.

For example, if γ = 2, then the modulating factor for an easy example (e.g., an example where the predicted probability of the correct class is 0.9) will be 0.01. This means that the loss for this example will be down-weighted by 99%.

On the other hand, the modulating factor for a hard example (e.g., an example where the predicted probability of the correct class is 0.1) will be 0.81. This means that the loss for this example will only be down-weighted by 19%.

Overall, focal loss allows the model to focus on learning from hard examples, which can improve the overall performance of the model.

In the context of this code, focal_loss(gamma=2) means that the model will use a focal loss function with a γ value of 2. This means that the model will focus on learning from hard examples more heavily.

Keep in mind, SparseCategoricalFocalLoss is a specialized loss function that is designed for sparse labels.



````python
# Train the model.
model.fit(x_train, y_train, epochs=10)
````

````python
# Evaluate the model on the test set.
test_loss, test_acc = model.evaluate(x_test, y_test)
print('Test accuracy:', test_acc)
````

# **Using Simple Convolutional Neural Networks (CNNs)**

````python
import tensorflow as tf
````

````python
# Load the MNIST dataset.
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()
````

````python
# Convert the data to float32 format.
x_train = tf.cast(x_train, tf.float32) / 255.0
x_test = tf.cast(x_test, tf.float32) / 255.0
````

The CNN model you provided has the following layers:

*   Convolutional layer: This layer is responsible for extracting features from the input images. It has 32 filters of size 3x3, which means that it will learn 32 different ways to represent the images.
*   Max pooling layer: This layer reduces the dimensionality of the data by downsampling it. It has a pool size of 2x2, which means that it will reduce the size of the input images by half.

*   Flatten layer: This layer converts the output of the pooling layer from a 3D tensor to a 1D tensor. This is necessary because the dense layers in the model can only handle 1D input tensors.
*   Dense layer: This layer is a fully connected layer. It takes the output of the flatten layer as input and produces a 128-dimensional output vector.
*   Dense layer: This layer is another fully connected layer. It takes the output of the previous dense layer as input and produces a 10-dimensional output vector. This is the final layer of the model, and its output represents the predicted probability of each class.

````python
# Define the model.
model = tf.keras.Sequential([
  # Convolutional layer with 32 filters of size 3x3.
  tf.keras.layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
  # Max pooling layer with a pool size of 2x2.
  tf.keras.layers.MaxPooling2D((2, 2)),
  # Flatten the output of the pooling layer.
  tf.keras.layers.Flatten(),
  # Dense layer with 128 units and a relu activation function.
  tf.keras.layers.Dense(128, activation='relu'),
  # Dense layer with 10 units and a softmax activation function.
  tf.keras.layers.Dense(10, activation='softmax')
])
````

````python
# Compile the model.
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])
````

````python
# Train the model.
model.fit(x_train, y_train, epochs=10)
````

````python
# Evaluate the model on the test set.
test_loss, test_acc = model.evaluate(x_test, y_test)
print('Test accuracy:', test_acc)
````
