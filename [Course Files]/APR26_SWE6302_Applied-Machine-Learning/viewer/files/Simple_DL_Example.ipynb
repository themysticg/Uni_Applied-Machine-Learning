This code will train a simple neural network to classify MNIST handwritten digits. The network has two hidden layers, each with 128 neurons and 10 neurons respectively. The output layer has 10 neurons, one for each digit (class).

Note: Use the code in Google Colab Notebook if possible

````python
#Import Tensorflow Library
import tensorflow as tf
````

````python
# Load the MNIST dataset.
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()
````

````python
# Convert the data to float32 format.
x_train = tf.cast(x_train, tf.float32) / 255.0
x_test = tf.cast(x_test, tf.float32) / 255.0
````

````python
# Create a sequential model. This is a slightly updated DL model with dropout layers
model = tf.keras.Sequential([
  tf.keras.layers.Flatten(input_shape=(28, 28)),

  # Add a dropout layer with a dropout rate of 20%.
  tf.keras.layers.Dropout(0.2),

  # Add a hidden layer with 128 neurons.
  tf.keras.layers.Dense(128, activation='relu'),

  # Add another dropout layer with a dropout rate of 20%.
  tf.keras.layers.Dropout(0.2),

  # Add a hidden layer with 64 neurons.
  tf.keras.layers.Dense(64, activation='relu'),

  # Add a final hidden layer with 10 neurons and a softmax activation for classification.
  tf.keras.layers.Dense(10, activation='softmax') #You can also use a different activation here such as activation='sigmoid' Note: softmax function converts a vector of real numbers into a probability distribution
])
````

````python
# Compile the model.
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])
'''
You can also try out different loss functions as follows to see how the loss changes such as:
loss='binary_crossentropy' OR
loss='categorical_crossentropy' OR
loss=focal_loss(gamma=2) #Focal loss
'''

'''
#By default optimizer used in Keras here, which is the Adam optimizer, has a default learning rate of 0.001. But the learning rate can be specified as follows.
model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.005),
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])
'''
````

````python
# Train the model.
model.fit(x_train, y_train, epochs=10)

'''
# To inlcude Batch Size try the follow ->
model.fit(x_train, y_train, epochs=10, batch_size=32)
'''
````

````python
# Evaluate the model on the test set.
test_loss, test_acc = model.evaluate(x_test, y_test)
print('Test accuracy:', test_acc)
````

# **Additional Notes (Working with Binary Crossentropy Loss Function)**

To use the binary_crossentropy loss function in your code, you need to ensure that the target labels (y_train and y_test) are binary or are in a format that can be interpreted for binary classification.

The MNIST dataset you are using has 10 classes (digits 0-9), which makes it a multi-class classification problem, not a binary one. Therefore, using binary_crossentropy directly would not work with the current setup because binary_crossentropy is intended for binary classification tasks (with labels of 0 or 1).

Key Modifications Needed:
- **Dataset**: Filter the dataset to include only two classes (0 and 1).
- **Output Layer**: Change the final output layer to have 1 neuron with sigmoid activation.
- **Loss Function**: Use binary_crossentropy in the model compilation.
This will allow the model to run with binary cross-entropy as the loss function.

````python
# Use the following code to leverage Binary Crossentropy Loss Function

# Filter the dataset to keep only 0s and 1s
binary_filter = (y_train == 0) | (y_train == 1)
x_train_binary = x_train[binary_filter]
y_train_binary = y_train[binary_filter]

binary_filter_test = (y_test == 0) | (y_test == 1)
x_test_binary = x_test[binary_filter_test]
y_test_binary = y_test[binary_filter_test]

# Modify the model for binary classification
model = tf.keras.Sequential([
  tf.keras.layers.Flatten(input_shape=(28, 28)),

  # Add a dropout layer with a dropout rate of 20%.
  tf.keras.layers.Dropout(0.2),

  # Add a hidden layer with 128 neurons.
  tf.keras.layers.Dense(128, activation='relu'),

  # Add another dropout layer with a dropout rate of 20%.
  tf.keras.layers.Dropout(0.2),

  # Add a hidden layer with 64 neurons.
  tf.keras.layers.Dense(64, activation='relu'),

  # Change the final output layer to 1 neuron with sigmoid activation for binary classification.
  # When using binary cross-entropy as the loss function, the reason we need to change the final output layer to 1 neuron with a sigmoid activation is due to the nature of binary classification and how binary cross-entropy works.
  # Binary classification involves two classes, typically labeled as 0 and 1. The sigmoid function is ideal for binary classification because it outputs a probability between 0 and 1. Specifically, it takes any real number as input and squashes it to a value between 0 and 1.
  # Binary cross-entropy measures the difference between the predicted probability (from the sigmoid output) and the true binary label (0 or 1).
  # In binary classification, we only need one output that represents the probability of the instance belonging to the positive class (class 1). The output neuron should have sigmoid activation so that the final output is a probability (between 0 and 1).
  tf.keras.layers.Dense(1, activation='sigmoid')
])

# Compile the model using binary_crossentropy.
model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])

# Train the model
model.fit(x_train_binary, y_train_binary, epochs=10)

# To inlcude Batch Size try the follow ->
# model.fit(x_train_binary, y_train_binary, epochs=10, batch_size=32)
````

````python
# Evaluate the model on the test set.
test_loss, test_acc = model.evaluate(x_test_binary, y_test_binary)
print('Test accuracy:', test_acc)
````

# **Additional Notes (Working with Categorical Crossentropy Loss Function)**

To modify the code to use categorical_crossentropy as the loss function, you need to convert the labels (y_train and y_test) to one-hot encoded format because categorical_crossentropy expects the labels to be in that format. In the current setup, you're using sparse_categorical_crossentropy, which works with integer labels directly.

Key Changes Needed:
- **Convert Labels**: The labels are converted from integers to one-hot encoded vectors using tf.keras.utils.to_categorical().
- **Loss Function**: The loss function is changed from sparse_categorical_crossentropy to categorical_crossentropy in the model.compile() method.
- **Training and Evaluation**: You now pass y_train_one_hot and y_test_one_hot (the one-hot encoded labels) when training and evaluating the model.

````python
# Use the following code to leverage Categorical Crossentropy Loss Function

# Load the MNIST dataset.
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

# Convert the data to float32 format.
x_train = tf.cast(x_train, tf.float32) / 255.0
x_test = tf.cast(x_test, tf.float32) / 255.0

# Convert the labels to one-hot encoded format
y_train_one_hot = tf.keras.utils.to_categorical(y_train, num_classes=10)
y_test_one_hot = tf.keras.utils.to_categorical(y_test, num_classes=10)

# Create a sequential model. This is a slightly updated DL model with dropout layers.
model = tf.keras.Sequential([
    tf.keras.layers.Flatten(input_shape=(28, 28)),

    # Add a dropout layer with a dropout rate of 20%.
    tf.keras.layers.Dropout(0.2),

    # Add a hidden layer with 128 neurons.
    tf.keras.layers.Dense(128, activation='relu'),

    # Add another dropout layer with a dropout rate of 20%.
    tf.keras.layers.Dropout(0.2),

    # Add a hidden layer with 64 neurons.
    tf.keras.layers.Dense(64, activation='relu'),

    # Add a final hidden layer with 10 neurons and a softmax activation for classification.
    tf.keras.layers.Dense(10, activation='softmax') # softmax is appropriate for multiclass classification
])

# Compile the model with categorical_crossentropy loss
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])

# Train the model.
model.fit(x_train, y_train_one_hot, epochs=10)

'''
# To inlcude Batch Size try the follow ->
model.fit(x_train, y_train, epochs=10, batch_size=32)
'''
````

````python
# Evaluate the model on the test set.
test_loss, test_acc = model.evaluate(x_test, y_test_one_hot)
print('Test accuracy:', test_acc)
````
