```python
import pandas as pd

# Create a disctionary of data
data = {
    'Duration':[30, 40, 45, 60, 75],
    'Average_Pulse':[80,85,90,100,120],
    'Max_Pulse': [120, 130, 140, 145, 150],
    'Calorie_Burnage': [240, 260, 290, 300, 330],
    'Hours_Work': [10, 8, 0, 7, 8],
    'Hours_Sleep': [7, 8, 7, 8, 8]
       }
```

```python
# Check the type of data
type(data)
```

```python
# Create a dataframe from the dictionary
df = pd.DataFrame(data)
df.head(5)
```

```python
import matplotlib.pyplot as plt

df.plot(x ='Average_Pulse', y='Calorie_Burnage', kind='line')
plt.ylim(ymin=0)
plt.xlim(xmin=0)
plt.show()
```

```python
# Build slope method on your own ->
def slope(x1, y1, x2, y2):
    s = (y2-y1)/(x2-x1)
    return s
print (slope(80,240,90,260))
```

```python
# Fetch the slope and intercept based on x, y values
import numpy as np
x = df['Average_Pulse']
y = df['Calorie_Burnage']

slope_intercept = np.polyfit(x,y,1)

print(slope_intercept)
type(slope_intercept)
```

```python
# Showcase the line based on slope and intercept
x = np.linspace(-2, 2, 100)
y = slope_intercept[0] * x + slope_intercept[1] 

fig = plt.figure(figsize = (10, 5))
plt.plot(x, y)
plt.show()
```

```python
# Building our own prediction
def pred(x):
    return slope_intercept[0] * x + slope_intercept[1] 

print(pred(100))
```

```python
# Build our own linear rgression model and showcase the data points on plots
from scipy import stats

x = df['Average_Pulse']
y = df['Calorie_Burnage']
m = slope_intercept[0]
C = slope_intercept[1]

mymodel = list(map(pred, x))
print(mymodel)

plt.scatter(x, y)
plt.plot(x, m * x + C)
#plt.ylim(ymin=0, ymax=2000)
#plt.xlim(xmin=0, xmax=200)
plt.xlabel("Average_Pulse")
plt.ylabel ("Calorie_Burnage")
plt.show()
```

```python
# Scatter plot

x = df['Average_Pulse']
y = df['Calorie_Burnage']

slope, intercept, r, p, std_err = stats.linregress(x, y)

print('Correlation coefficient: ' + str(r))

def mypred(x):
    return slope * x + intercept
mymodel = list(map(mypred, x))

plt.scatter(x, y)
plt.plot(x, mymodel)
plt.xlabel("Average_Pulse")
plt.ylabel ("Calorie_Burnage")
plt.show()
```
