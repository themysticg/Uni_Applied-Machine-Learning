```python
# Import statistics Library
import statistics as st

# Calculate average values
print(st.mean([1, 3, 5, 7, 9, 11, 13]))
```

```python
# Calculate median value
data1 = [1, 3, 5, 7, 9, 10, 11, 13, 16, 17]
data2 = [1, 3, 5, 7, 10, 11, 13, 16, 17]
print(len(data1))
print(len(data2))
print(st.median(data1))
print(st.median(data2))
```

```python
# Calculate mode value
data = [0, 2, 3, 2, 1, 0, 0, 2, 0, 0]
print(st.mode(data))
```

```python
# Calculate variance value
data = [1, 4, 5, 4, 8]
print(st.variance(data))
```

```python
# Calculate std value
data = [1, 4, 5, 4, 8]
print(st.mean(data))
std = st.stdev(data)
print(std)
4/std
```

```python
# Calculate the probability that a random value is less than or equal to 40
# load packages
from scipy.stats import norm

prob1 = norm.cdf(x=40, loc=50, scale=10)
print(prob1)
```

```python
# load packages
from statistics import NormalDist

NormalDist(mu=50, sigma=10).pdf(40)
```

```python
# Calculate the probability that a random value is greater than or equal to 40
# load packages
from scipy.stats import norm

prob1 = 1 - norm.cdf(x=40, loc=50, scale=10)
print(prob1)
```

```python
# Calculate the probability that a random value that is between 400 and 500. Nrmally distributed data with a mean of 200 and a standard deviation of 50.
# load packages
from scipy.stats import norm

norm.cdf(x=500, loc=200, scale=50) - norm.cdf(x=400, loc=200, scale=50)
```

```python
# Import necessary libraries
import pandas as pd

# Numerical Data (Discrete and Continuous)
numerical_data = {
"Heights": [165, 170, 168, 172, 177], # Continuous
"Purchases": [2, 3, 1, 5, 4] # Discrete
}

# Categorical Data
categorical_data = {
"Gender": ["Male", "Female", "Female", "Male", "Male"],
"State": ["TX", "CA", "NY", "CA", "TX"]
}

# Ordinal Data
ordinal_data = {
"Movie Rating": [5, 3, 4, 2, 1] # 1 to 5 scale
}

# Display data types
print("Numerical Data:", numerical_data)
print("Categorical Data:", categorical_data)
print("Ordinal Data:", ordinal_data)
```

```python
import numpy as np
from scipy import stats
data = [0, 2, 3, 2, 1, 0, 0, 2, 0]

# Calculate Mean
mean_value = np.mean(data)
print("Mean:", mean_value)

# Calculate Median
median_value = np.median(data)
print("Median:", median_value)

# Calculate Mode
mode_value = stats.mode(data)
print("Mode:", mode_value)

# Calculate Variance
variance_value = np.var(data)
print("Variance:", variance_value)

# Calculate Standard Deviation
std_dev_value = np.std(data)
print("Standard Deviation:", std_dev_value)
```

```python
import matplotlib.pyplot as plt
# Histogram for the dataset
plt.hist(data, bins=5, edgecolor='black')
plt.title("Histogram of Dataset")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```
