# Predict the CO2 emission of a car based on the size of the engine

```python
import pandas as pd
```

```python
df = pd.read_csv('/My Work/DataScience/York_university/YSA_course/Semester_Sept/Course_presentation/Session 3/Activity/MultipleRegressionData.csv')
df.head(5)
```

```python
X = df[['Volume', 'Weight']]
X.head()
```

```python
X.tail()
```

```python
X = X.fillna(0)
X.tail()
```

```python
y = df.CO2
y.tail()
```

```python
y= y.fillna(0)
y.tail()
```

```python
from sklearn import linear_model
```

```python
reg = linear_model.LinearRegression()
```

```python
reg.fit(X,y)
```

#predict the CO2 emission of a car where the weight is 2300kg, and the volume is 1300cm3

```python
predictedCO2 = reg.predict([[2300, 1300]])
```
