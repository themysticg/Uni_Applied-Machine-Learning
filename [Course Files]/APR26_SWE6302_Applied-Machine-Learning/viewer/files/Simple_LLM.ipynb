Simple LLM using LSTM ::->

```python
from tensorflow.keras.datasets import reuters
from tensorflow.keras.preprocessing.sequence import pad_sequences
import tensorflow as tf
from tensorflow.keras import layers

# Load and pad the data with a specified vocabulary size
vocab_size = 10000
(x_train, y_train), (x_test, y_test) = reuters.load_data(num_words=vocab_size)
max_length = 200
x_train = pad_sequences(x_train, maxlen=max_length)
x_test = pad_sequences(x_test, maxlen=max_length)

# Define the model
encoder = tf.keras.Sequential([
    layers.Embedding(input_dim=vocab_size, output_dim=128),  # Adjust input_dim to vocab_size
    layers.LSTM(128, return_sequences=True),
    layers.LSTM(64, return_sequences=True)
])

decoder = tf.keras.Sequential([
    layers.LSTM(64, return_sequences=True),
    layers.LSTM(32),
    layers.Dense(46, activation='softmax')
])

model = tf.keras.Sequential([
    encoder,
    decoder
])

# Compile the model
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])

# Train the model
model.fit(x_train, y_train, epochs=10, validation_split=0.1)

# Evaluate the model
loss, accuracy = model.evaluate(x_test, y_test)  # This will now return both loss and accuracy
print('Loss:', loss)
print('Accuracy:', accuracy)
```

In the context of processing textual data with neural networks, particularly in natural language processing (NLP) tasks, **max_length = 200** refers to the maximum length of the sequences (like sentences or documents) that the model will process. The `layers.Embedding` layer in a neural network model, particularly in the context of natural language processing (NLP), serves as a crucial component for handling textual data. This layer transforms integer-encoded categorical data (like word indices) into dense vectors of fixed size. Each integer (representing a word) is mapped to a vector in a higher-dimensional space, allowing the model to capture more information about each word than the integer encoding alone. For example, in a dataset where words are represented by unique integers, the embedding layer will convert each integer into a dense vector. These vectors are learned during the training process, enabling the model to understand and capture the relationships and semantic meanings of words. By doing so, the embedding layer facilitates more nuanced and effective processing of textual data, significantly enhancing the model's ability to perform tasks like text classification, sentiment analysis, or language translation.

In the `layers.Embedding` component of a neural network model, the parameter `output_dim=64` specifies the size of the output vectors for each word. When you set `output_dim` to 64, it means that each word in the input data (represented as an integer index) will be converted into a 64-dimensional vector by the embedding layer. These dense vectors capture the semantic properties of words in a multi-dimensional space, where each dimension contributes to representing the word's meaning and context in a way that the model can understand and utilize. The choice of 64 as the dimensionality is a balance between model complexity and expressiveness: higher dimensions can capture more nuanced semantics but increase computational complexity, while lower dimensions are computationally efficient but may not capture as much detail. This vector representation enables the neural network to perform operations on words, recognizing patterns and relationships in the text data more effectively than with the original integer encodings.

Improving the validation accuracy of a neural network, especially on a complex dataset like Reuters, can be challenging and often requires experimentation with various aspects of the model architecture, data preprocessing, and training process. Here are several strategies you can try to enhance the performance of your model:

### 1. Model Architecture Adjustments:
- **Bidirectional LSTM:** Using bidirectional LSTMs can help the model capture context from both past and future data points in a sequence. A Bidirectional LSTM (BiLSTM) is a type of recurrent neural network (RNN) that extends the traditional LSTM (Long Short-Term Memory) model by adding a second layer that processes the input sequence in reverse order. The key idea behind BiLSTMs is to provide the network with complete past and future context for any point in the input sequence.
- **Dropout:** Adding dropout layers can help prevent overfitting by randomly setting a fraction of input units to 0 during training.
- **Regularization:** Apply L1 or L2 regularization in dense layers to penalize large weights and help prevent overfitting. L1 and L2 regularization are techniques used in machine learning to prevent overfitting, where a model performs well on training data but poorly on unseen data. Both are forms of regularization, which add a penalty to the loss function used during training. This penalty encourages simpler models, potentially reducing overfitting.

### 2. Data Preprocessing and Feature Engineering:
- **Word Embeddings:** Instead of using a simple embedding layer, consider using pre-trained embeddings like GloVe or Word2Vec, which can provide a richer representation of words.
- **Sequence Length:** Experiment with different `max_length` values. If sequences are truncated too much, important information might be lost.

### 3. Training Process:
- **Learning Rate Scheduling:** Adjust the learning rate over time (e.g., reduce it when the validation accuracy plateaus).
- **Early Stopping:** Use early stopping to terminate training when the validation accuracy stops improving.
- **Batch Size and Epochs:** Experiment with different batch sizes and numbers of epochs. Smaller batch sizes can sometimes offer more stable and generalizable learning.


### Final Thoughts:
- **Experimentation:** Finding the right combination of techniques and parameters often requires trial and error. Be prepared to experiment and iteratively refine your model.
- **Validation Data:** Ensure that the validation data is representative of the problem space and is not biased in any way.
- **Performance Metrics:** Accuracy is just one metric. Depending on your task, other metrics like precision, recall, or F1-score might also be important to consider.

Slightly more updated LLM using LSTM ::->

```python
from tensorflow.keras.datasets import reuters
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras import layers, models, regularizers
import tensorflow as tf

# Load and pad the data
vocab_size = 10000
(x_train, y_train), (x_test, y_test) = reuters.load_data(num_words=vocab_size)
max_length = 200
x_train = pad_sequences(x_train, maxlen=max_length)
x_test = pad_sequences(x_test, maxlen=max_length)

# Define the model
# Encoder
encoder = tf.keras.Sequential(name='Encoder', layers=[
    layers.Embedding(input_dim=vocab_size, output_dim=64),
    layers.Bidirectional(layers.LSTM(64, return_sequences=True)),
    layers.GlobalAveragePooling1D()
])

# Decoder
decoder = tf.keras.Sequential(name='Decoder', layers=[
    layers.Dense(64, activation='relu', kernel_regularizer=regularizers.l2(0.001)),
    layers.Dropout(0.5),
    layers.Dense(46, activation='softmax')
])

model = tf.keras.Sequential([
    encoder,
    decoder
])

# Compile the model
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])

# Early Stopping Callback
early_stopping = tf.keras.callbacks.EarlyStopping(monitor='val_accuracy', patience=5)

# Train the model
model.fit(x_train, y_train, epochs=20, batch_size=128, validation_split=0.1, callbacks=[early_stopping])

# Evaluate the model
loss, accuracy = model.evaluate(x_test, y_test)
print('Loss:', loss)
print('Accuracy:', accuracy)
```

LLM using Transformer (creating your own simple model) ::->

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.layers import Dense, Embedding, LayerNormalization, MultiHeadAttention, Dropout, Input, Lambda
from tensorflow.keras.models import Model, Sequential
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.losses import SparseCategoricalCrossentropy
from tensorflow.keras.datasets import reuters
from tensorflow.keras.preprocessing.sequence import pad_sequences


# Parameters
vocab_size = 10000  # Size of the vocabulary
max_length = 200  # Length of input sequences
embed_dim = 64  # Embedding size
num_heads = 2  # Number of attention heads
ff_dim = 64  # Hidden layer size in feed forward network

# Load and pad the Reuters dataset
(x_train, _), (x_test, _) = reuters.load_data(num_words=vocab_size)
x_train = pad_sequences(x_train, maxlen=max_length, padding='post')
x_test = pad_sequences(x_test, maxlen=max_length, padding='post')

# Preparing target data for text generation
y_train = x_train[:, 1:] # Shift the input sequences one token to the right
y_test = x_test[:, 1:]

# Transformer block as a layer
class TransformerBlock(tf.keras.layers.Layer):
    def __init__(self, embed_dim, num_heads, ff_dim):
        super(TransformerBlock, self).__init__()
        self.att = MultiHeadAttention(num_heads=num_heads, key_dim=embed_dim)
        self.ffn = Sequential([Dense(ff_dim, activation="relu"), Dense(embed_dim)])
        self.layernorm1 = LayerNormalization(epsilon=1e-6)
        self.layernorm2 = LayerNormalization(epsilon=1e-6)
        self.dropout1 = Dropout(0.1)
        self.dropout2 = Dropout(0.1)

    def call(self, inputs, training):
        attn_output = self.att(inputs, inputs)
        attn_output = self.dropout1(attn_output, training=training)
        out1 = self.layernorm1(inputs + attn_output)
        ffn_output = self.ffn(out1)
        ffn_output = self.dropout2(ffn_output, training=training)
        return self.layernorm2(out1 + ffn_output)

# Building the model
inputs = Input(shape=(max_length,))
embedding_layer = Embedding(input_dim=vocab_size, output_dim=embed_dim, input_length=max_length)(inputs)
transformer_block = TransformerBlock(embed_dim, num_heads, ff_dim)(embedding_layer)

# Slice the output to match the target sequence length
transformer_block_sliced = Lambda(lambda x: x[:, :-1, :])(transformer_block)

outputs = Dense(vocab_size, activation="softmax")(transformer_block_sliced)
model = Model(inputs=inputs, outputs=outputs)

# Compile the model
model.compile(optimizer=Adam(1e-4), loss='sparse_categorical_crossentropy', metrics=["accuracy"])

# Training the model
model.fit(x_train, y_train, epochs=15, batch_size=32, validation_split=0.1)

# Evaluate the model
loss, accuracy = model.evaluate(x_test, y_test)
print('Loss:', loss)
print('Accuracy:', accuracy)

# Save the entire model to a file
model.save('my_llm_transformer_model.keras')
```
