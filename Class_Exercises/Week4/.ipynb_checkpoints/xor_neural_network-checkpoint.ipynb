# Week 4: Neural Network for the XOR Problem

## Step 1 — Import NumPy and Create the Dataset

This notebook implements a multilayer neural network from scratch using NumPy.

The network learns the XOR logic gate using forward propagation and backpropagation.

XOR returns `1` when its two inputs are different and returns `0` when they are the same.

````python
# Import NumPy for arrays, matrix multiplication
# and other numerical calculations.
import numpy as np

# Set a random seed so that the randomly generated
# starting weights remain reproducible.
np.random.seed(42)

# Confirm that NumPy loaded correctly.
print("NumPy imported successfully.")
````

````python
# Create the four possible input combinations
# for an XOR logic gate.
#
# Each row represents one training example.
# Each column represents one input feature.
X = np.array([
    [0, 0],  # Both inputs are equal
    [0, 1],  # The inputs are different
    [1, 0],  # The inputs are different
    [1, 1]   # Both inputs are equal
], dtype=float)

# Create the correct XOR output for every input.
#
# XOR returns 1 when the inputs differ.
# XOR returns 0 when the inputs are equal.
y = np.array([
    [0],  # 0 XOR 0 = 0
    [1],  # 0 XOR 1 = 1
    [1],  # 1 XOR 0 = 1
    [0]   # 1 XOR 1 = 0
], dtype=float)

# Display the dataset.
print("XOR inputs:")
print(X)

print("\nExpected outputs:")
print(y)

# Check the dimensions of both arrays.
print("\nInput shape:", X.shape)
print("Output shape:", y.shape)
````

## Step 2 — Define the Neural Network

The neural network contains three layers:

- **Input layer:** receives the two XOR input values.
- **Hidden layer:** contains four neurons that learn non-linear patterns.
- **Output layer:** contains one neuron that produces the final prediction.

The network starts with random weights and zero biases. Training will gradually adjust these values.

````python
# Define a class representing our neural network.
class NeuralNetwork:

    # The constructor runs when a NeuralNetwork object is created.
    def __init__(self, input_size, hidden_size, output_size):

        # Store the number of neurons in each layer.
        self.input_size = input_size
        self.hidden_size = hidden_size
        self.output_size = output_size

        # Create random weights connecting the input layer
        # to the hidden layer.
        #
        # For this network, the shape is:
        # 2 input neurons × 4 hidden neurons = (2, 4)
        self.weights_input_hidden = np.random.randn(
            self.input_size,
            self.hidden_size
        )

        # Create random weights connecting the hidden layer
        # to the output layer.
        #
        # For this network, the shape is:
        # 4 hidden neurons × 1 output neuron = (4, 1)
        self.weights_hidden_output = np.random.randn(
            self.hidden_size,
            self.output_size
        )

        # Create one zero bias for every hidden neuron.
        # The shape is (1, 4).
        self.bias_hidden = np.zeros(
            (1, self.hidden_size)
        )

        # Create one zero bias for the output neuron.
        # The shape is (1, 1).
        self.bias_output = np.zeros(
            (1, self.output_size)
        )

    # Define the sigmoid activation function.
    def sigmoid(self, x):

        # Convert any input value into a value between 0 and 1.
        return 1 / (1 + np.exp(-x))

    # Define the derivative of the sigmoid function.
    def sigmoid_derivative(self, sigmoid_output):

        # Measure the slope of the sigmoid function.
        #
        # Backpropagation uses this value to determine
        # how strongly each parameter should change.
        return sigmoid_output * (1 - sigmoid_output)


# Confirm that the class-definition cell ran successfully.
print("NeuralNetwork class structure defined successfully.")
````

## Step 3 — Forward Propagation

Forward propagation passes the input data through the network to produce predictions.

The process has two stages:

1. Calculate and activate the hidden-layer values.
2. Use the hidden-layer outputs to calculate the final prediction.

Matrix multiplication combines the inputs with the learned weights. Bias values are then added before applying the sigmoid activation function.

````python
# Define a class representing our neural network.
class NeuralNetwork:

    # The constructor creates the network's weights and biases.
    def __init__(self, input_size, hidden_size, output_size):

        # Store the number of neurons in each layer.
        self.input_size = input_size
        self.hidden_size = hidden_size
        self.output_size = output_size

        # Create random weights connecting the input layer
        # to the hidden layer.
        #
        # Shape for this exercise:
        # 2 input neurons × 4 hidden neurons = (2, 4)
        self.weights_input_hidden = np.random.randn(
            self.input_size,
            self.hidden_size
        )

        # Create random weights connecting the hidden layer
        # to the output layer.
        #
        # Shape for this exercise:
        # 4 hidden neurons × 1 output neuron = (4, 1)
        self.weights_hidden_output = np.random.randn(
            self.hidden_size,
            self.output_size
        )

        # Create one zero bias for each hidden neuron.
        self.bias_hidden = np.zeros(
            (1, self.hidden_size)
        )

        # Create one zero bias for the output neuron.
        self.bias_output = np.zeros(
            (1, self.output_size)
        )

    # Define the sigmoid activation function.
    def sigmoid(self, x):

        # Convert values into the range from 0 to 1.
        return 1 / (1 + np.exp(-x))

    # Define the derivative of the sigmoid function.
    def sigmoid_derivative(self, sigmoid_output):

        # Calculate the slope required during backpropagation.
        return sigmoid_output * (1 - sigmoid_output)

    # Define the forward-propagation process.
    def feedforward(self, X):

        # Calculate the hidden layer's weighted sums.
        #
        # X has shape (4, 2).
        # weights_input_hidden has shape (2, 4).
        # The resulting hidden_activation has shape (4, 4).
        self.hidden_activation = (
            np.dot(X, self.weights_input_hidden)
            + self.bias_hidden
        )

        # Apply sigmoid activation to every hidden-layer value.
        self.hidden_output = self.sigmoid(
            self.hidden_activation
        )

        # Calculate the output layer's weighted sums.
        #
        # hidden_output has shape (4, 4).
        # weights_hidden_output has shape (4, 1).
        # The resulting output_activation has shape (4, 1).
        self.output_activation = (
            np.dot(
                self.hidden_output,
                self.weights_hidden_output
            )
            + self.bias_output
        )

        # Apply sigmoid activation to produce probabilities
        # between 0 and 1.
        self.predicted_output = self.sigmoid(
            self.output_activation
        )

        # Return the network's predictions.
        return self.predicted_output

    # Update the network by propagating its error backwards.
    def backward(self, X, y, learning_rate):

        # Measure the difference between the expected outputs
        # and the network's current predictions.
        output_error = y - self.predicted_output

        # Calculate the output-layer adjustment signal.
        output_delta = (
            output_error
            * self.sigmoid_derivative(self.predicted_output)
        )

        # Send the output error backwards to the hidden layer.
        hidden_error = np.dot(
            output_delta,
            self.weights_hidden_output.T
        )

        # Calculate the hidden-layer adjustment signal.
        hidden_delta = (
            hidden_error
            * self.sigmoid_derivative(self.hidden_output)
        )

        # Update the weights and biases between the hidden
        # layer and output layer.
        self.weights_hidden_output += (
            np.dot(self.hidden_output.T, output_delta)
            * learning_rate
        )
        self.bias_output += (
            np.sum(output_delta, axis=0, keepdims=True)
            * learning_rate
        )

        # Update the weights and biases between the input
        # layer and hidden layer.
        self.weights_input_hidden += (
            np.dot(X.T, hidden_delta)
            * learning_rate
        )
        self.bias_hidden += (
            np.sum(hidden_delta, axis=0, keepdims=True)
            * learning_rate
        )

    # Train the network over several complete passes
    # through the XOR dataset.
    def train(self, X, y, epochs, learning_rate):

        # Store the learning progress for later graphs.
        self.loss_history = []
        self.accuracy_history = []

        for epoch in range(epochs):
            # Produce predictions with the current parameters.
            output = self.feedforward(X)

            # Measure the mean squared prediction error.
            loss = np.mean(np.square(y - output))

            # Convert probabilities to binary predictions and
            # calculate the current classification accuracy.
            binary_predictions = (output >= 0.5).astype(int)
            accuracy = np.mean(binary_predictions == y)

            self.loss_history.append(loss)
            self.accuracy_history.append(accuracy)

            # Improve the weights and biases.
            self.backward(X, y, learning_rate)

            # Report progress without printing every epoch.
            if (
                epoch == 0
                or (epoch + 1) % 2000 == 0
                or epoch == epochs - 1
            ):
                print(
                    f"Epoch {epoch + 1:5d} | "
                    f"Loss: {loss:.6f} | "
                    f"Accuracy: {accuracy * 100:.1f}%"
                )

        return self


# This line is outside the class because it has no indentation.
print("Complete NeuralNetwork class defined successfully.")
````

````python
# Create a network with:
# 2 input neurons, 4 hidden neurons and 1 output neuron.
nn = NeuralNetwork(
    input_size=2,
    hidden_size=4,
    output_size=1
)

# Pass the XOR inputs through the untrained network.
initial_predictions = nn.feedforward(X)

# Display the predictions produced by random starting weights.
print("Predictions before training:")
print(initial_predictions)

# Display the prediction-array dimensions.
print("\nPrediction shape:", initial_predictions.shape)
````

## Step 4 — Backpropagation

Backpropagation trains the network by working backwards from the output layer.

The network:

1. Compares its prediction with the expected output.
2. Calculates the output-layer gradient.
3. Propagates the error back into the hidden layer.
4. Updates the weights and biases to reduce future errors.

The learning rate controls the size of each update.

> **Reference:** This method has been merged into the complete `NeuralNetwork` class above.

```python
    # Define the backpropagation process.
    def backward(self, X, y, learning_rate):

        # Calculate the difference between the expected outputs
        # and the network's current predictions.
        output_error = y - self.predicted_output

        # Calculate the output-layer adjustment signal.
        #
        # The error is multiplied by the sigmoid derivative
        # because the derivative indicates how sensitive the
        # output is to changes in its weighted input.
        output_delta = (
            output_error
            * self.sigmoid_derivative(
                self.predicted_output
            )
        )

        # Send the output error backwards through the
        # hidden-to-output weights.
        hidden_error = np.dot(
            output_delta,
            self.weights_hidden_output.T
        )

        # Calculate the hidden-layer adjustment signal.
        hidden_delta = (
            hidden_error
            * self.sigmoid_derivative(
                self.hidden_output
            )
        )

        # Update the weights connecting the hidden layer
        # to the output layer.
        #
        # hidden_output.T changes the hidden output from
        # shape (4, 4) to shape (4, 4).
        # Multiplying by output_delta produces a (4, 1)
        # adjustment matching weights_hidden_output.
        self.weights_hidden_output += (
            np.dot(
                self.hidden_output.T,
                output_delta
            )
            * learning_rate
        )

        # Update the output neuron's bias.
        #
        # Summing across axis 0 combines the adjustments
        # from all four training examples.
        self.bias_output += (
            np.sum(
                output_delta,
                axis=0,
                keepdims=True
            )
            * learning_rate
        )

        # Update the weights connecting the input layer
        # to the hidden layer.
        #
        # X.T has shape (2, 4).
        # hidden_delta has shape (4, 4).
        # Their dot product has shape (2, 4).
        self.weights_input_hidden += (
            np.dot(
                X.T,
                hidden_delta
            )
            * learning_rate
        )

        # Update the four hidden-neuron biases.
        self.bias_hidden += (
            np.sum(
                hidden_delta,
                axis=0,
                keepdims=True
            )
            * learning_rate
        )
```

Backpropagation now belongs to the complete class definition above.

> **Reference:** This method has been merged into the complete `NeuralNetwork` class above.

```python
    # Train the neural network over multiple epochs.
    def train(self, X, y, epochs, learning_rate):

        # Create empty lists to record training progress.
        self.loss_history = []
        self.accuracy_history = []

        # Repeat the complete learning process.
        for epoch in range(epochs):

            # Perform forward propagation using the
            # network's current weights and biases.
            output = self.feedforward(X)

            # Calculate the mean squared error before
            # updating the network's parameters.
            loss = np.mean(
                np.square(y - output)
            )

            # Convert probabilities into binary predictions.
            binary_predictions = (
                output >= 0.5
            ).astype(int)

            # Calculate the proportion of correct predictions.
            accuracy = np.mean(
                binary_predictions == y
            )

            # Save the current loss and accuracy.
            self.loss_history.append(loss)
            self.accuracy_history.append(accuracy)

            # Update the weights and biases using
            # backpropagation.
            self.backward(
                X,
                y,
                learning_rate
            )

            # Display progress at selected intervals.
            #
            # epoch + 1 produces human-friendly numbers
            # beginning at 1 instead of 0.
            if (
                epoch == 0
                or (epoch + 1) % 2000 == 0
                or epoch == epochs - 1
            ):
                print(
                    f"Epoch {epoch + 1:5d} | "
                    f"Loss: {loss:.6f} | "
                    f"Accuracy: {accuracy * 100:.1f}%"
                )

        # Return the trained network object.
        return self
```

The complete class now includes forward propagation, backpropagation and training.

## Step 5 — Train the Neural Network

The network will train for 10,000 epochs using a learning rate of `0.1`.

During each epoch, the network performs forward propagation, calculates its prediction error and uses backpropagation to update its weights and biases.

````python
# Reset the random seed so the experiment
# produces reproducible starting parameters.
np.random.seed(42)

# Create a fresh neural network containing:
# 2 input neurons, 4 hidden neurons and 1 output neuron.
nn = NeuralNetwork(
    input_size=2,
    hidden_size=4,
    output_size=1
)

# Train the network on the XOR dataset.
nn.train(
    X=X,
    y=y,
    epochs=10000,
    learning_rate=0.1
)
````

## Step 6 — Test the Trained Neural Network

The trained network now processes all four XOR inputs again.

The output neuron returns a probability between `0` and `1`. A threshold of `0.5` converts each probability into a binary class:

- Values below `0.5` become class `0`.
- Values of `0.5` or higher become class `1`.

The predicted classes should match the expected XOR outputs.

````python
# Perform one final forward pass using the
# network's trained weights and biases.
final_probabilities = nn.feedforward(X)

# Convert the output probabilities into binary classes.
#
# A probability of 0.5 or higher becomes 1.
# A probability below 0.5 becomes 0.
final_predictions = (
    final_probabilities >= 0.5
).astype(int)

# Print a heading for the results table.
print("Trained XOR neural network")
print("-" * 65)
print("x1 | x2 | Probability | Prediction | Expected | Result")
print("-" * 65)

# Compare every prediction with its expected answer.
for inputs, probability, prediction, expected in zip(
    X,
    final_probabilities,
    final_predictions,
    y
):
    # Extract the two input values.
    x1, x2 = inputs

    # Extract ordinary scalar values from the arrays.
    probability_value = float(probability[0])
    predicted_class = int(prediction[0])
    expected_class = int(expected[0])

    # Check whether the network produced the correct answer.
    result = (
        "Correct"
        if predicted_class == expected_class
        else "Incorrect"
    )

    # Display one row of the results table.
    print(
        f"{int(x1):>2} | "
        f"{int(x2):>2} | "
        f"{probability_value:>11.4f} | "
        f"{predicted_class:>10} | "
        f"{expected_class:>8} | "
        f"{result}"
    )

# Calculate the final classification accuracy.
final_accuracy = np.mean(final_predictions == y)

# Display the overall result as a percentage.
print("-" * 65)
print(f"Final accuracy: {final_accuracy * 100:.1f}%")
````

## Step 7 — Visualise the Training Progress

The training history shows how the network improved over 10,000 epochs.

- **Loss** measures the difference between the expected outputs and predicted probabilities. It should decrease during training.
- **Accuracy** measures the proportion of correctly classified XOR examples. It should eventually reach 100%.

The graphs help confirm that backpropagation successfully trained the network.

````python
# Import Matplotlib for creating graphs.
import matplotlib.pyplot as plt

# Create epoch numbers beginning at 1.
epoch_numbers = range(
    1,
    len(nn.loss_history) + 1
)

# Create one figure containing two graphs.
#
# 1 row and 2 columns place the graphs side by side.
figure, axes = plt.subplots(
    1,
    2,
    figsize=(14, 5)
)

# Plot the training loss on the first graph.
axes[0].plot(
    epoch_numbers,
    nn.loss_history,
    color="steelblue",
    linewidth=2
)

# Add labels and a title to the loss graph.
axes[0].set_title("XOR Neural Network Training Loss")
axes[0].set_xlabel("Epoch")
axes[0].set_ylabel("Mean Squared Error")

# Add a light grid to make the values easier to read.
axes[0].grid(
    True,
    linestyle="--",
    alpha=0.5
)

# Plot the training accuracy on the second graph.
#
# Multiply each decimal value by 100 to display percentages.
accuracy_percentages = [
    accuracy * 100
    for accuracy in nn.accuracy_history
]

axes[1].plot(
    epoch_numbers,
    accuracy_percentages,
    color="seagreen",
    linewidth=2
)

# Add labels and a title to the accuracy graph.
axes[1].set_title("XOR Neural Network Training Accuracy")
axes[1].set_xlabel("Epoch")
axes[1].set_ylabel("Accuracy (%)")

# Keep the vertical accuracy scale between 0% and 105%.
axes[1].set_ylim(0, 105)

# Add a light grid.
axes[1].grid(
    True,
    linestyle="--",
    alpha=0.5
)

# Adjust spacing so that titles and labels do not overlap.
plt.tight_layout()

# Display both graphs.
plt.show()
````
